```python
import numpy as numpy
import pandas as pd
```

```python
credits = pd.read_csv('tmdb_5000_credits.csv')
movie = pd.read_csv('tmdb_5000_movies.csv')
```

MERGED THE DATASETS ON THE BASIS OF TITLE


```python
movies = movie.merge(credits,on='title')
```

```python
movies = movies[['id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.dropna(inplace=True)
```

SEPARATING OUT GENRES

Developing a Helper Function to get genres as an ARRAY

```python
import ast
```

```python
def convert(obj):#defines a function named convert obj = one cell value from the genres column
    L = []#Creates an empty list
    for i in ast.literal_eval(obj):#Converts the string into a real Python list
        L.append(i['name'])#Takes only the 'name' value
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

FOR CAST TAKING ONLY THE FIRST 3 NAMES

```python
def convert3(obj):
    L = []
    count =0
    for i in ast.literal_eval(obj):
        if count!=3:
         L.append(i['name'])#Takes only the 'name' value
         count+=1
        else:
           break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
         L.append(i['name'])#Takes only the 'name' value
         break
    return L
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

REMOVE THE SPACES IN THE NAMES(TO AVOID CONFUSION)

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['tags']=movies['overview'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[['id','title','tags']]
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
```

FOR STEMMING (['LOVED','LOVING']) --> (['LOVE','LOVE'])

```python
!pip install nltk
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

CountVectorizer converts text data → numbers...
Keeps only the top 5000 most frequent words....
Removes common words like:
the, is, am, are, and, in

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

GOT THE MOST RECURRING WORDS


```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title']==movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)),reverse = True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```
